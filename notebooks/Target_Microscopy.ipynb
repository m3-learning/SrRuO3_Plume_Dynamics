```python
# --- Combined overview across scans and ROIs ---
combined_sample_dict = {
    'Background': [
        [(0.0, 0.0), (830.0, 0.0), (830.0, 830.0), (0.0, 830.0)],
    ],
    'G1': [
        [(5.0, 5.0), (830.0, 5.0), (830.0, 830.0), (5.0, 830.0)],
        [(5.0, 5.0), (400.0, 5.0), (410.0, 350.0), (5.0, 200.0)],
        [(500.0, 400.0), (830.0, 500.0), (830.0, 830.0), (550.0, 830.0)],
    ],
    'G2_': [
        [(5.0, 5.0), (830.0, 5.0), (830.0, 830.0), (5.0, 830.0)],
        [(5.0, 650.0), (150.0, 400.0), (280.0, 830.0), (5.0, 830.0)],
        [(400.0, 500.0), (830.0, 500.0), (830.0, 830.0), (400.0, 830.0)],
    ],
    'G3': [
        [(5.0, 5.0), (830.0, 5.0), (830.0, 830.0), (5.0, 830.0)],
        [(5.0, 5.0), (340.0, 5.0), (360.0, 220.0), (5.0, 830.0)],
        [(520.0, 5.0), (830.0, 5.0), (830.0, 400.0), (450.0, 400.0)],
    ],
    'G4': [
        [(5.0, 5.0), (830.0, 5.0), (830.0, 830.0), (5.0, 830.0)],
        [(5.0, 310.0), (830.0, 210.0), (830.0, 830.0), (5.0, 830.0)],
        [(5.0, 5.0), (830.0, 5.0), (830.0, 180.0), (5.0, 270.0)],
    ],
    'G5': [
        [(5.0, 5.0), (830.0, 5.0), (830.0, 830.0), (5.0, 830.0)],
        [(5.0, 180.0), (830.0, 80.0), (830.0, 830.0), (5.0, 830.0)],
        [(5.0, 5.0), (500.0, 5.0), (500.0, 30.0), (5.0, 100.0)],
    ],
}

combined_surfaces = []
for name, rois in combined_sample_dict.items():
    for idx, roi in enumerate(rois, start=1):
        combined_surfaces.append({
            'datx': f"../data/TargetMicroscopy/selected/{name}.datx",
            'txt': f"../data/TargetMicroscopy/selected/{name}.txt",
            'polygon': roi,
            'label': f"{name}-roi{idx}",
        })

combined_results, combined_fig = mic.analyze_surfaces(
    combined_surfaces,
    polygon_unit='um',
    roughness_unit='um',
    xy_unit='um',
    z_unit='um',
    max_cols=3,
)

for res in combined_results:
    print(f"{res['label']}: Rq = {res['Rq']:.3f} µm, Ra = {res['Ra']:.3f} µm, Rz = {res['Rz']:.3f} µm, mean = {res['mean']:.3f} µm")
combined_results
```

```python
import matplotlib.pyplot as plt

labels = [res['label'] for res in combined_results]
metric_names = ['Ra', 'Rq', 'Rz', 'mean']

fig, axes = plt.subplots(len(metric_names), 1,
                         figsize=(max(8, len(labels) * 0.7), 3 * len(metric_names)),
                         sharex=True)

x_vals = range(len(labels))

for ax, metric in zip(axes, metric_names):
    values = [res[metric] for res in combined_results]
    ax.plot(x_vals, values, marker='o')
    ax.set_ylabel(f"{metric} (µm)")
    ax.grid(alpha=0.3)
    ax.set_title(f"{metric} across scans/ROIs")

axes[-1].set_xticks(list(x_vals))
axes[-1].set_xticklabels(labels, rotation=45, ha='right')
axes[-1].set_xlabel("ROI label")

fig.tight_layout()
plt.show()
```

```python
from sro_sto_plume import microscopy as mic
from pathlib import Path

base = Path("../data/TargetMicroscopy")         # adjust if notebook lives elsewhere
datx_path = base / "G1.datx"
txt_path = base / "G1.txt"

height_map, geom = mic.load_processed_surface(datx_path, txt_path)
stats_full = mic.compute_roughness(height_map, geometry=geom, unit="um")
stats_full
```

```python
corner_um = [(0, 0), (70, 70)]
stats_roi = mic.compute_roughness(
    height_map,
    geometry=geom,
    unit="um",
    corners=corner_um,
)
stats_roi
```

```python
pixel_slice = (slice(100, 200), slice(300, 450))
stats_pixel = mic.compute_roughness(
    height_map,
    pixel_region=pixel_slice,
    unit="um",
)
stats_pixel
```
