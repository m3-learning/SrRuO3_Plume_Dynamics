# 4. Plume Metrics Extraction

```python
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt
import numpy as np
from plume_learn.plume_analyzer.analyze_function import load_plumes_and_align, run_plume_analysis

plume_recording_root = '../data/Plumes/plume_recordings/'
coords_root = '../data/Plumes/frame_normalize_dataset/'
output_root = '../data/Plumes/plume_metrics/'
```

### 1. G1 - YG065

```python
ds_name, ds_id = 'G1', 'YG065'
file_path = plume_recording_root + 'YG065_YichenGuo_09102024.h5'
coords_path = coords_root + 'YG065_coords.npy'
standard_coords_path = coords_root + 'standard_coords.npy'
output_csv_path = output_root + 'YG065_metrics.csv'

frame_view_index = 2
plumes, frame_view, fig = load_plumes_and_align(file_path, frame_view_index=frame_view_index)

coords = np.array([[34,51], [36,215], [382,47], [382,229]])
ds_metric = {'ds_id':ds_id, 'ds_name':ds_name}
viz_parms = {'viz':True, 'index':frame_view_index, 'viz_index':list(np.arange(0, 24, 1)), 'plume_name': ds_name, 'progress_bar':False}
# metric_parms = {'threshold_list':[200, 5, 'flexible'], 'rename_dataset':True} # 500ns; 54mm/347pixels
metric_parms = {'time_interval':500e-9, 'meter_per_pixel': 54/1000/347, 'threshold_list':[200, 5, 'flexible'], 'rename_dataset':True, 'skip_empty_plumes':True} # 500ns; 54mm/347pixels
align_parms = {'align': True, 'frame_view': frame_view, 'coords':coords, 'coords_path': coords_path, 'standard_coords_path': standard_coords_path}
run_plume_analysis(plumes, output_csv_path, align_parms=align_parms, ds_metric=ds_metric, viz_parms=viz_parms, metric_parms=metric_parms)        
```
